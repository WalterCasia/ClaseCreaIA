## Conceptos claves de HTTP

HTTP (*Hypertext Trasfer Protocol*)

es un protocolo de solicitud-restapuesta para comunicacion entre clientes (navegador, app movil, script) y servidores

### Froma general de un apeticion
```
<Metodo><Ruta> HTTP/1.1
Host:ejemplo.com
Header-1: valor
Header-2: valor
```

##Metodo HTTP (CRUD)
- **GET** = leer recursos (Sin cambiar estado)
- **POST** = Crear recursos (cambia de estado)
- **PUT** = reemplazar un recurso completo (actualizar)
- **PATCH** = modifica parcialmente un recuros
- **DELETE** = Borrar un recurso.

## IDEMPOTENCIA

una operacion es idempotente si ejecutar una vez o varias veces termian en le mismo estado de servidor.
- **IDEMPOTENTES**: GET, PUT, DELETE
- **NO IDEMPOTENTE: POST (cada POST podria crear un nuveo recurso)

### CODIGOS DE ESTADO MAS COMUNES

400 Bad Request: El servidor no procesa la solicitud por una sintaxis inválida.
401 Unauthorized: La página requiere autenticación previa mediante credenciales válidas.
403 Forbidden: El servidor entiende la petición pero rechaza dar acceso por falta de permisos.
404 Not Found: El recurso solicitado no existe o la URL está mal escrita.
405 Method Not Allowed: El método de petición (como POST o GET) no está permitido para el recurso.
408 Request Timeout: El servidor cerró la conexión porque el cliente tardó demasiado en responder.
500 Internal Server Error: Fallo genérico en el servidor sin un motivo específico detallado.
502 Bad Gateway: El servidor actuó como intermediario y recibió una respuesta inválida del servidor principal.
503 Service Unavailable: El servidor no puede responder por sobrecarga temporal o mantenimiento técnico.
504 Gateway Timeout: El servidor intermediario no recibió una respuesta a tiempo del servidor de origen. 


## Encabezados de Solicitud (Enviados por el Cliente)
Host: Especifica el dominio del servidor al que se envía la petición.
User-Agent: Identifica el navegador, sistema operativo y dispositivo que realiza la consulta.
Accept: Indica los formatos de contenido que el cliente puede procesar (ej. text/html, application/json).
Authorization: Contiene las credenciales para autenticar al usuario ante el servidor (ej. un token JWT).
Cookie: Envía las cookies guardadas previamente por el servidor de vuelta a este.
Encabezados de Respuesta (Enviados por el Servidor)
Content-Type: Define el tipo de medio (MIME) del recurso retornado (ej. application/json, text/html; charset=utf-8).
Content-Length: Indica el tamaño del cuerpo de la respuesta en bytes.
Server: Informa sobre el software del servidor web que procesó la solicitud (ej. Apache, nginx).

````python
from http.server import BaseHTTPRequestHandler, HTTPServer
import json

class MyHandler(BaseHTTPRequestHandler):
    def send_json(self, data):
        self.send_response(200)
        self.send_header('Content-Type', 'application/json')
        self.end_headers()
        body = json.dumps(data).encode('utf-8')
        self.wfile.write(body)

    def do_GET(self):
        if self.path == '/':
            self.send_json({'message': 'Hola Mundo desde la raiz'})
        elif self.path == '/hello':
            self.send_json({'message': 'Hello World from /hello'})
        else:
            self.send_response(404)
            self.end_headers()
            self.wfile.write(b'Not Found')

if __name__ == '__main__':
    server = HTTPServer(('localhost', 8080), MyHandler)
    print('Servidor HTTP escuchando en http://localhost:8080')
    server.serve_forever()
````

# Express
es un framework minimalista simplica la crecacion de sevidores HTTP

## QUE ES UN MIDDLEWARE

En express, un middleware es una funcuin que se ejecuta entre la peticion (request) y la respuesta response.

**capa intermedia que procesa la peticion o la respuesta**

````python
// Estructra de middleware
// req: la peticion del cliente
// res: la respuesta del servidor
// next(): funcion que llamaal siguiente middleware en la cadena


function miMiddleware(req, res, next) {
    console.log("Metodo", req.metod, "Ruta", req.url);
    next();
}

app.use(miMiddleware)
````

````python

const express = require('express');
const app = express();
const port = 3000;

app.use(express.json());

app.post("/usuraio", (req, res) => {
    console.log(req.body);
    res.json({ mensaje: "Usuario recibido", datos: req.body});
})

app.listen(port, () => {console.log("Servidore escucahndo en puerto")})
````
