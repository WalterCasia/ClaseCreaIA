## Que es HTML?
**(HyperText Markup Languaje)**
- es un lenguaje estamdar para definir la **escritura** de paginas web
- Una pagina HTML es un **arbol** de elementos: una raiz <html>
nodos , hijos;
- <head>
- <body>


## Estructura base de un documento
- iniciar con `<!DOCTYPE HTML>` (indica que es HTML) y contine la raiz <html> con dos secciones:
    - <head></head> : metadatos (titulos, estilos, scripts, idioma, etc)
    - <body></body> : contenido visible

````python
<!DOCTYPE html>
<html lang="es">    
    <head>
        <title>Mi primera pagina web</title>
    </head>
    <body>
        <h1>Hola mundo</h1>
        <p>Esta es mi primera pagina web</p>
    </body>
</html>
````

## Texto
- **Titulo** <h1> al <h6>, de mayor a menor importancia
- **Parrafos** <p> para bloques de texto.
- **Saltos de linea>**: <br>
- **Separadores** : <hr> para una linea horizontal


````python
<!DOCTYPE html>
<html lang="es">    
    <head>
        <title>Mi primera pagina web</title>
    </head>
    <body>
        <header>
            <h1>Bienvenido a mi pagina web</h1>
        </header>
        <main>
            <p>Estoy aprendiendo a crear paginas web</p>
            <p>Esta es la seccion del main</p>
        </main>
        <footer>
            <p>Derechos reservados 2026</p>
        </footer>

    </body>
</html>
````

- valor: el nuevo aspecto que le daremos a esa propiedad.
